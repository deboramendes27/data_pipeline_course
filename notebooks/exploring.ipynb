## Trabalhando com CSV

```python
path_csv = "../data_raw/dados_empresaB.csv"
```

```python
import csv

dados_csv = []  
with open(path_csv, "r") as f:
    spamreader = csv.reader(f, delimiter=',')
    for row in spamreader:
        dados_csv.append(row)
    
```

```python
dados_csv_tratados = []
for row in dados_csv[1:]:
    line_dict = {}
    for i in range(len(row)):
        line_dict[dados_csv[0][i]] = row[i]
    dados_csv_tratados.append(line_dict)
```

para fazer o mesmo, podemos apenas usar a função DictReader

```python
dados_csv = []  
with open(path_csv, "r") as f:
    spamreader = csv.DictReader(f, delimiter=',')
    for row in spamreader:
        dados_csv.append(row)
```

## Comparando os dados

```python
columns_csv = list(dados_csv[0].keys())
columns_csv
```

```python
columns_json = list(dados_json[0].keys())
columns_json
```

Os nomes escolhidos são dos arquivos em json

```python
key_mapping = {'Nome do Item': 'Nome do Produto',
                'Classificação do Produto': 'Categoria do Produto',
                'Valor em Reais (R$)': 'Preço do Produto (R$)',
                'Quantidade em Estoque': 'Quantidade em Estoque',
                'Nome da Loja': 'Filial',
                'Data da Venda': 'Data da Venda'}
key_mapping
```

```python
new_dados_csv = []

for dicts in dados_csv:
    dict_temp = {}
    for old_key, value in dicts.items():
        dict_temp[key_mapping[old_key]] = value
    new_dados_csv.append(dict_temp)

new_dados_csv[0]
```

## Juntando as listas

```python
len(dados_json), len(new_dados_csv)
```

```python
combined_list = dados_json + new_dados_csv
# Poderia usar combined_list.extend(new_dados_csv) também
```

```python
len(combined_list)
```

```python
combined_list[0]
```

```python
nome_colunas = list(combined_list[-1].keys())
nome_colunas
```

```python
path_combined = "../data_processed/dados_combinados.csv"

with open(path_combined, "w") as f:
    writer = csv.DictWriter(f, fieldnames=nome_colunas)
    writer.writeheader()

    for row in combined_list:
        writer.writerow(row)
```
